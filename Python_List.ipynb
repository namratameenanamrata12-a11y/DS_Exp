# List
## List is oredered collection of iteams stored in []
### mutable
### oredered
### duplicates are allowed

```python
# create a list
marks = [98,87,97,58,77]
print(marks)
```

```python
student_info=[101,"Namrata","CSE",7.04]
print(student_info)
```

```python
# access the element using index
name=["Namrata","Avdesh",'Meena','Jaat']
print(name)
print(name[2])
print(name[1])
print(name[-1])
```

```python
# example for mutable int the list
cgpa=[7.04,9,9.3,9.9]
print(cgpa)
cgpa[0]=8.5
print("After updation")
print(cgpa)
```

```python
# add an item in the list
Name=["Namrata","Avdesh"]
Name.append("JAAT")
print(Name)
```

```python
# insert in particular index
student_name=["Namrata","Avdesh","kanha","shagun","reena"]
student_name.insert(1,"Jaat")
print(student_name)
```

```python
# extend function
sec_a=["Namrata","Avdesh","JAAt"]
sec_b=["Kanha","Shagun","Reena"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
# remove an item in the list
sec_a_students=["Namrata","Avdesh","reena","Meena","Jaat"]
sec_a_students.remove("Meena")
print(sec_a_students)
del sec_a_students[2]
print(sec_a_students)
sec_a_students.pop()
print(sec_a_students)
```

```python
# list slicing
subject = ["Phython",
           "statistics",
           "java","Os",
           "physics","CN"]
print(subject[1:4])
print(subject[:3])
print(subject[4:])
print(subject[::2])
# length list
print("Length of the list:",len(subject))
```

```python
marks=[99,92,97,98,95,91]
marks.sort()
print(marks)
# sort in decesing order
marks.sort(reverse=True)
print(marks)
```

```python
# nested list
marks=[
    [91,96,88],
    [78,98,76],
    [56,98,68,89]
    ]
print(marks)
print(marks[1])
print(marks[2][1])
```
