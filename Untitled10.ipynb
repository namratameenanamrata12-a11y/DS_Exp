```python
import numpy as np
import pandas as pd
```

```python
data={
    'Student_Id':[101,102,103,104,105,106],
    'Name':['Rahul Kumar', 'Amit sharma','Priya singh','Neha verma','Ravi kumar','ritika'],
    'Gender':['Male','Male','Female','Female','Male','female'],
    'Department':['CSE','CSE','ECE','ME','CSE',None],
    'Age':[91,20,21,18,22,None],
    'Attendaence':[90,85,96,97,98,92],
    'Marks':[85,99,98,96,97,None],
    'Email':['rahul@gmail.com','amit@gamil.com','priya@gmail.com','neha@gamil.com','ravi@gmail.com',None]
}
df = pd.DataFrame(data)
display(df)
```

```python
# Basic Inspection
print("Shape:",df.shape)
print("Columns:",df.columns)
print("Data Types:",df.dtypes)
display(df.head())
display(df.tail(3))
```

```python
# Detect the missing values
print(df.isnull().sum())
print("Total Missing Value:",df.isnull().sum().sum())
display(df[df.isnull().any(axis=1)]) # finds row with at least one missing value
```

```python
# Handle the missing value
df['Age'] = df['Age'].fillna(df['Age'].mean())
display(df)
```
